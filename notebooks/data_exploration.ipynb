#  Dataset Exploration & Initial Data Validation
**Project:** Historical Weather India (2000-2024)  
**Objective:** Inspect, validate data quality, and export the initial clean baseline copy.

```python
import os
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Construct the relative path from notebooks/ directory to data/raw/
file_path = os.path.join("..", "data", "raw", "india_2000_2024_daily_weather.csv")

# Load dataset
df = pd.read_csv(file_path)

# Quick structural verification
print(f"Dataset successfully loaded from: {file_path}")
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

```python
print("=== FIRST FIVE ROWS ===")
print(df.head())

print("\n=== LAST FIVE ROWS ===")
print(df.tail())
```

##  Understand Every Column

| Column Name | Meaning | Data Type | Expected Type |
| :--- | :--- | :--- | :--- |
| `city` | Name of the Indian city | object | object |
| `date` | Daily observation timestamp | object | datetime64 |
| `temperature_2m_max` | Maximum air temperature at 2 meters | float64 | float64 |
| `temperature_2m_min` | Minimum air temperature at 2 meters | float64 | float64 |
| `apparent_temperature_max` | Feels-like max temperature | float64 | float64 |
| `apparent_temperature_min` | Feels-like min temperature | float64 | float64 |
| `precipitation_sum` | Total daily precipitation (rain + liquid equivalent) | float64 | float64 |
| `rain_sum` | Total daily rainfall amount | float64 | float64 |
| `weather_code` | WMO weather interpretation code | int64 | int64 |
| `wind_speed_10m_max` | Maximum wind speed at 10 meters | float64 | float64 |
| `wind_gusts_10m_max` | Maximum wind gusts at 10 meters | float64 | float64 |
| `wind_direction_10m_dominant` | Dominant wind direction in degrees (0-360°) | int64 | int64 |

```python
# Check 1 & 2: Dimensions and explicit structural types
print("=== DATA TYPES AND NON-NULL COUNDS ===")
print(df.info())

# Check 3: Missing values count per column
print("\n=== MISSING VALUES PER COLUMN ===")
print(df.isnull().sum())

# Check 4: Exact duplicate row count
print(f"\nTotal Duplicate Records found: {df.duplicated().sum()}")
```

```python
# Check 5: Unique cities (up to first 20)
unique_cities = df['city'].unique()
print("=== UNIQUE CITIES ===")
print(f"Total Unique Cities Count: {len(unique_cities)}")
print(f"Cities List: {unique_cities[:20]}")

# Check 6: Exact Date Ranges
print("\n=== DATE RANGE VALIDATION ===")
print(f"Earliest Recorded Date: {df['date'].min()}")
print(f"Latest Recorded Date:   {df['date'].max()}")
```

```python
# Check 7 & Target 6: Descriptive Statistical Summary
print("=== STATISTICAL SUMMARY FOR VALUE INSPECTION ===")
print(df.describe())
```

```python
# Define target destination path
output_dir = os.path.join("..", "data", "processed")
output_file = os.path.join(output_dir, "weather_clean_v1.csv")

# Create directory if it does not exist
os.makedirs(output_dir, exist_ok=True)

# Export copy without modifying raw content or keeping original system index
df.to_csv(output_file, index=False)
print(f"Successfully saved un-modified copy to baseline: {output_file}")
```
